```python
import pandas as pd
```

```python
df = pd.read_csv('student_performance.csv')
```

```python
df.head()
```

```python
print("The Total numer of students:",df.shape[0])
print("The total number of columns:",df.shape[1])
print("The total number of department:",len(df['department'].unique()))
```

```python
print(df['department'].value_counts())
```

```python
print("Highest math score",df['math_score'].max())
print("Lowest math score",df['math_score'].min())
print("Average math score",df['math_score'].mean().round(2))
print("Average programming score:",df['programming_score'].mean().round(2))
print("Average Attendance:",df['attendance_percentage'].mean(),"%")
```

```python
low_att=df[df['attendance_percentage']<75][['name','department','attendance_percentage']]
if len(low_att)>0:
  print(low_att.to_string(index=False))

else:
  print("All students have attendanace above 75%")
```
